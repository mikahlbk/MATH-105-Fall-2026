## What did the company encode for us?

The original columns are still included. For modeling, use the prepared **numeric** columns:

| Column | Meaning |
|---|---|
| `Distance_scaled` | Distance in kilometers divided by 10 |
| `Prep_scaled` | Preparation time in minutes divided by 10 |
| `Experience_scaled` | Courier experience in years divided by 10 |
| `Traffic_Medium`, `Traffic_High` | 1 if that traffic level, 0 otherwise; **Low** is the reference |

**Why divide by 10?** To help gradient descent behave more predictably. When reporting predictions and RMSE, multiply scaled delivery times by 10 to return to **minutes**. When interpreting coefficients, explain the units and what was scaled.

## Exercise 19 — Interpret the coefficients

Because we divided the numerical inputs and delivery time by 10, **do not interpret the scaled weights as minutes per kilometer/minute/year without converting them**.

For example, if `Distance_scaled = Distance_km / 10` and `Delivery_scaled = Delivery_Time_min / 10`, the corresponding weight is in **minutes per kilometer**. The same cancellation applies to preparation time (minutes per minute) and experience (minutes per year).

For a 0/1 indicator (traffic or time of day), multiply its fitted weight by 10 to interpret the predicted **difference in delivery minutes** relative to the reference category, holding other inputs fixed.

Use this table to support your explanation on the handout.
