## What did the company encode for us?

Fast-ish Foods has provided some additional information about each delivery. But there's a small problem: some of this information is recorded as words rather than numbers!

For example, traffic conditions are recorded as **Low, Medium, or High**. How can we include these categories in a mathematical model?

One approach is to create new columns containing only 0s and 1s. These are called **indicator variables** (or dummy variables).

### Example: Traffic conditions

Instead of one column describing traffic, we create two new columns:

| Traffic | Traffic_Medium | Traffic_High |
|---|---|---|
| Low | 0 | 0 |
| Medium | 1 | 0 |
| High | 0 | 1 |

Each new column answers a yes-or-no question:
- `Traffic_Medium`: Is traffic medium? 1 = yes, 0 = no.
- `Traffic_High`: Is traffic high? 1 = yes, 0 = no.

**Why only two columns when there are three categories?**

Because we can already identify Low traffic when both columns are 0! We don't need a third column.

Low traffic is called our **reference category**.

### What about time of day?

The same idea works for time of day, which has four categories:

| Time of Day | Time_Afternoon | Time_Evening | Time_Night |
|---|---|---|---|
| Morning | 0 | 0 | 0 |
| Afternoon | 1 | 0 | 0 |
| Evening | 0 | 1 | 0 |
| Night | 0 | 0 | 1 |

Here, **Morning** is our reference category.

### What about the other variables?

The remaining variables are already numerical. Fast-ish Foods divided some of them by 10 to make the values easier to work with during gradient descent.

| Column | Meaning |
|---|---|
| `Distance_scaled` | Distance (km) divided by 10 |
| `Prep_scaled` | Preparation time (minutes) divided by 10 |
| `Experience_scaled` | Courier experience (years) divided by 10 |
| `Traffic_Medium`, `Traffic_High` | Indicators for traffic; Low is the reference |
| `Time_Afternoon`, `Time_Evening`, `Time_Night` | Indicators for time of day; Morning is the reference |
| `Delivery_scaled` | Delivery time (minutes) divided by 10 |

**Why divide by 10?** This helps gradient descent work with parameter updates on more manageable numerical scales.

When reporting predicted delivery times and RMSE, remember to multiply the scaled results by 10 to return to **minutes**.
