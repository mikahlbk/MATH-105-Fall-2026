### Question 2. Reading the log-loss graph (7 points)

Suppose an email **really is spam**, so $y=1$. The model predicts $\hat y$, the probability that it is spam.

For this observation, the two costs are

$$\text{Squared Error}=(1-\hat y)^2, \qquad \text{Cross-Entropy}=-\ln(\hat y).$$

Run the graph code below. It compares the two costs for different predictions. **The graph is provided; you do not need to create it.**

```python
# Provided graph — run this cell. You do not need to edit it.
p = np.linspace(0.01, 0.99, 400)
plt.figure(figsize=(7, 4))
plt.plot(p, -np.log(p), label='Cross-entropy: -ln(p)')
plt.plot(p, (1 - p)**2, label='Squared error: (1-p)^2', linestyle='--')
plt.xlabel('Predicted probability of spam (p)')
plt.ylabel('Cost for an email that IS spam (y = 1)')
plt.title('How much does each cost penalize a prediction?')
plt.ylim(0, 5)
plt.grid(alpha=0.25)
plt.legend()
plt.show()
```

We want to train the model

$$\hat y=\frac{1}{1+e^{-(mx+b)}}$$

by minimizing its **average binary cross-entropy**.

The starter code below already constructs the cost function and calculates its partial derivatives using SymPy. **You only need to complete the two gradient descent update lines.**

**(a) [4 pts]** Complete the code to update $m$ and $b$ using gradient descent. Use the provided learning rate $\eta=0.2$, initial values $m=b=0$, and 200 iterations.

**(b) [1 pt]** Run the code. Report the final values of $m$, $b$, and the cost.

```python
# Provided setup: no edits needed here.
x = np.array([0, 1, 2, 3, 4, 5], dtype=float)
y = np.array([0, 0, 0, 1, 1, 1], dtype=float)
N = len(x)

m, b = sp.symbols('m b')
y_hat = [1 / (1 + sp.exp(-(m * xi + b))) for xi in x]
C = -(1 / N) * sum(
    yi * sp.log(pi) + (1 - yi) * sp.log(1 - pi)
    for yi, pi in zip(y, y_hat)
)

dC_dm = sp.diff(C, m)
dC_db = sp.diff(C, b)
grad_m = sp.lambdify((m, b), dC_dm, 'numpy')
grad_b = sp.lambdify((m, b), dC_db, 'numpy')
cost = sp.lambdify((m, b), C, 'numpy')

print(f"Starting cross-entropy: {cost(0.0, 0.0):.3f}")
```
