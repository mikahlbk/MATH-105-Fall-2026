## Part 2: Estimate the mysterious limit

The expression

(2^h - 1)/h

is the slope of the secant line connecting (0, 1) and (h, 2^h). As h approaches 0, this secant slope should approach the slope of the tangent line at (0, 1).


### Make a prediction

Look at the graph below. What do you predict about the slope of the tangent line at (0, 1)? Is it closer to 0, 0.5, 1, or 2? Explain your prediction.


```python
import numpy as np
import matplotlib.pyplot as plt

x_values = np.linspace(-3, 3, 400)
y_values = 2**x_values

plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.plot(x_values, y_values, label='f(x) = 2^x')
plt.scatter([0], [1], color='red', zorder=3, label='(0, 1)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('The graph of f(x) = 2^x')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

### Use secant lines to estimate the tangent slope

The code below computes

m = [f(0+h) - f(0)] / h = (2^h - 1)/h

Run the code for several positive and negative values of h.


```python
# Define the function
f = lambda x: 2**x

# Values of h to investigate
h_values = [1, 0.1, 0.01, 0.001, 0.0001,
            -1, -0.1, -0.01, -0.001, -0.0001]

# Compute the secant slope from x = 0 to x = h
for h in h_values:
    slope = (f(0 + h) - f(0)) / h
    print(f'h = {h:>8}: secant slope = {slope:.6f}')
```

## Part 4: Check the pattern at other points

Now we will estimate the derivative directly at several values of x using

f'(x) approximately equal to [f(x+h) - f(x)] / h

Before running the code, predict: What do you think the ratio f'(x)/f(x) will be?


```python
f = lambda x: 2**x
h = 0.0001

for x in [0, 1, 2, 3, 5]:
    derivative_estimate = (f(x + h) - f(x)) / h
    ratio = derivative_estimate / f(x)

    print(f'x = {x}: f(x) = {f(x):.4f}, '
          f"f'(x) approx {derivative_estimate:.4f}, "
          f"f'(x)/f(x) approx {ratio:.4f}")
```

### Discuss the pattern

## Final takeaway

Our numerical evidence suggests that

d/dx (2^x) is approximately 0.693 * 2^x

The exact constant is ln(2), and ln(2) is approximately 0.693. Therefore:

d/dx (2^x) = ln(2) * 2^x

More generally:

d/dx (a^x) = ln(a) * a^x
