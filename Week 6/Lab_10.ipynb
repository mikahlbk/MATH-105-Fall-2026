### Part 1: Apply the Definition of the Derivative

The derivative of the function $f$ at the point $x$ is defined by:

$$f'(x) = \lim_{h \to 0} \frac{f(x+h)-f(x)}{h}.$$

Assume $f(x) = e^.5x$. Write an expression for $f'(x)$ using the definition. You do not need to evaluate the limit.


```python
#Write your answer here or on a separate sheet of paper.
```

Using properties of exponents we are going to simplify this expression.

$$f'(x) = \lim_{h \to 0} e^{.5x}\left (\frac{2^{.5h} - 1}{h} \right).$$

Notice that the limit inside the big parentheses does not depend on $x$. If we estimate that one limit, we can use it to find the derivative at any $x$ value.


### Part 2: Estimate the mystery limit from the graph

We can think of the expression

$$\frac{e^{.5h} - 1}{h}$$

as the slope of the secant line connecting $(0, 1)$ and $(h, e^.5h)$. As $h$ approaches $0$, this secant slope should approach the slope of the tangent line to the graph at the point $(0, 1)$.

Look at the graph below. What do you predict about the slope of the tangent line at $(0, 1)$? Is it closer to 0, 0.5, 1, or 2? Explain your prediction.


```python
import numpy as np
import matplotlib.pyplot as plt

#set up array of input values
x_values = np.linspace(-3, 3, 400)

#plot the secant line for a given value of h
h = 2
slope = lambda h:(np.exp(.5*h)-1)/h
secant_line_func= lambda x:slope(h)*(x-0)+2**0
plt.plot(x_values, secant_line_func(x_values), linestyle = '--',label='Secant Line')

#plot f(x) = e^.5x
exponential_func = lambda x: np.exp(.5*x)
plt.plot(x_values, exponential_func(x_values), label=r'$f(x) = e^{.5x}$')

#plot the point (0,1)
plt.scatter([0], [1], color='red', zorder=3, label='(0, 1)')

#graph properties
plt.ylim(-2,10)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid()
plt.legend()
plt.show()
```

### Part 3: Extimate the mystery limit numerically

You know the drill!

```python
# Define the function
f = lambda x: np.exp(.5*x)

# Values of h to investigate, left and right side
h_values = [1, 0.1, 0.01, 0.001, 0.0001,-1, -0.1, -0.01, -0.001, -0.0001]

# Compute the average rate of change from x = 0 to x = h
for h in h_values:
    slope = (f(0 + h) - f(0)) / h
    print("When h = {}, the averate rate of change = {}".format(h,slope))
```
