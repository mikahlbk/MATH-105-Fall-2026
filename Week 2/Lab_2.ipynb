We used Python to plot this data and thought about the choices we can make to help our visualization communicate effectively.



````python
# Import the plotting library
import matplotlib.pyplot as plt

# Enter the data
year = [1980, 1985, 1990, 1995, 2000, 2005, 2010, 2015, 2020, 2025]

avg_price = [.84, .80, 1.01, .92, .91, 1.22, 1.66, 2.47, 1.51, 5.11]

adjusted_price = [3.71, 3.44, 3.18, 2.99, 2.70, 3.31, 3.37, 3.38, 2.78, 5.11]

# Create the graph
plt.figure(figsize=(8, 5))

# Plot both sets of data as scatter
plt.scatter(year, avg_price, color='red', label="Average Price")
plt.scatter(year, adjusted_price, color='blue', label="Adjusted Price")

# Plot both sets of data as lines; don't need the labels again
plt.plot(year, avg_price, color='red')
plt.plot(year, adjusted_price, color='blue')

# Need this to add the labels, otherwise they won't show up
plt.legend()

# Change the horizontal and vertical axis limits
plt.xlim(1975, 2030)
plt.ylim(0, 10)

# Add labels and a title
plt.xlabel("Year")
plt.ylabel("Price ($)")
plt.title("Egg Prices Over Time")

plt.show()
````

Next, we're going to learn how to solve equations in Python! To do this, we'll introduce a new Python library called SymPy, which allows us to work with symbols and solve mathematical equations.

````python
#Import the sympy library
import sympy as sp

#Define symbol for input variable
t = sp.Symbol('t')

#Define formula
avg_price = 0.041*t-80.2

#Solve equation
solution_eggs = sp.solve(sp.Eq(avg_price, 8), t)

#Print solution
print("Solutions to AvgPrice(t) = 8:", solution_eggs)
````
