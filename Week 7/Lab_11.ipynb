We’ll use the following four-step process whenever we want to understand a function through its derivative.
    
1. Compute the derivative.
2. Find the critical points by solving f′(t) = 0 (or observing where the derivative may not exist).
3. Classify each critical point using the sign of the derivative.
4. Verify the results by graphing.

For example, consider the function
$$f(t) = \frac{1}{2}t^2-3t+2.$$

Step 1. Compute the Derivative.

```python
import sympy as sp

# Define the input variable
t = sp.Symbol('t')

# Define the function
f = 0.5*t**2 - 3*t + 2

# Compute the derivative
f_prime = sp.diff(f, t)

print("f(t) = ", f)
print("f'(t) = ", f_prime)
```

Step 2. Find the Critical Points.

```python
#set derivative equal to 0 and solve for t
critical_points = sp.solve(f_prime, t,0)

#answer is stored in variable called critical_points
print("Critical Points:", critical_points)
```

Step 3. Classify the Critical Points.

```python
#evaluate the derivative to the left of t = 3, I chose t = 2
print("Left Side: ", f_prime.subs(t,2))
#evaluate the derivative to the right of t = 3, I chose t = 4
print("Right Side: ", f_prime.subs(t,4))
```
