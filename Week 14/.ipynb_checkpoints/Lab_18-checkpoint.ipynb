## Part 1 · Explore waiting-time data (about 10 minutes)

If events arrive randomly at a constant average rate, the **exponential distribution** models the time until the next event. The rate is

$$\lambda=\frac{1}{5}=0.2\text{ requests per minute.}$$

The `scale=5` argument means the average waiting time is 5 minutes. Run this starter code to simulate 400 waiting times.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad

rng = np.random.default_rng(105)
rate = 1 / 5                           # requests per minute
wait_times = rng.exponential(scale=5, size=400)

print('First 12 simulated waiting times (minutes):')
print(np.round(wait_times[:12], 2))
print('Number of requests:', len(wait_times))
```

## Part 2 · A histogram whose areas represent probabilities (about 12 minutes)

We can show the waiting times with a histogram. In a **density histogram**, each bar's **area** represents a proportion, not its height.

Run the next cell. The bars show the simulated waiting times, and the smooth curve is the exponential probability density

$$f(t)=0.2e^{-0.2t},\qquad t\ge0.$$

The shaded region is the area between 0 and 3 minutes.

```python
# The code is provided. Run it; no edits needed.
width = 2
upper_edge = np.ceil(wait_times.max() / width) * width + width
edges = np.arange(0, upper_edge + 0.001, width)
heights, edges = np.histogram(wait_times, bins=edges, density=True)

def density(t):
    return rate * np.exp(-rate * t)

t = np.linspace(0, 25, 400)
t_shaded = np.linspace(0, 3, 100)
fig, ax = plt.subplots(figsize=(9, 4))
ax.hist(wait_times, bins=edges, density=True, alpha=0.6,
        label='Simulated waiting times (density histogram)')
ax.plot(t, density(t), linewidth=2, label='Exponential density model')
ax.fill_between(t_shaded, density(t_shaded), alpha=0.25,
                label='Model area: 0 to 3 minutes')
ax.set_xlim(0, 25)
ax.set_xlabel('Time until next request (minutes)')
ax.set_ylabel('Probability density (per minute)')
ax.set_title('Waiting times: simulated data and a mathematical model')
ax.legend()
plt.show()

first_area = heights[0] * (edges[1] - edges[0])
print(f'First bar: width = {edges[1]-edges[0]:.1f}, height = {heights[0]:.3f}')
print(f'Area of first bar = {first_area:.3f}')
print(f'Proportion of waits less than 2 minutes = {np.mean(wait_times < 2):.3f}')
print(f'Total area of histogram bars = {np.sum(heights * np.diff(edges)):.3f}')
```

## Part 3 · Riemann sums versus the integral (about 18 minutes)

Now calculate the area under the **smooth model** from 0 to 3 minutes:

$$P(0\le T\le3)=\int_0^3 0.2e^{-0.2t}\,dt.$$

The starter function `left_sum(n)` estimates this area using `n` equal-width left-endpoint rectangles. Try 6 and 60 rectangles. Then compare with the value returned by `quad`, a numerical integration tool.

```python
# The code for a left-endpoint Riemann sum is provided.
def left_sum(n):
    delta_t = 3 / n
    left_endpoints = np.linspace(0, 3, n, endpoint=False)
    return np.sum(density(left_endpoints) * delta_t)

# TODO 3: Call left_sum with 6 rectangles and with 60 rectangles.
estimate_6 = np.nan
estimate_60 = np.nan

# Python can also calculate the integral directly.
integral_value, integration_error = quad(density, 0, 3)

print(f'6 rectangles (left):  {estimate_6:.3f}')
print(f'60 rectangles (left): {estimate_60:.3f}')
print(f'Integral from 0 to 3: {integral_value:.3f}')
print(f'Simulated fraction:   {proportion_3_or_less:.3f}')
```

## Part 4 · Change the rate (about 10 minutes)

Suppose requests now arrive at an average rate of **one every 10 minutes**, rather than one every 5 minutes.

**Predict before running:** Would the probability of a request arriving within 3 minutes be higher or lower? Why?

Then use the provided function to calculate the new probability.

```python
# ANSWER 4a (prediction): I expect the probability to be ... because ...
#

def probability_within_three(event_rate):
    result, _ = quad(lambda t: event_rate * np.exp(-event_rate * t), 0, 3)
    return result

# TODO 4: Replace np.nan with the new average rate in requests/minute.
new_rate = np.nan

original_probability = probability_within_three(rate)
new_probability = probability_within_three(new_rate) if not np.isnan(new_rate) else np.nan

print(f'One request every 5 minutes:  {original_probability:.3f}')
print(f'One request every 10 minutes: {new_probability:.3f}')

# ANSWER 4b: Did the result match your prediction? Explain in context.
#
```
