### Using Python to Evaluate a Limit

So far, we've approximated instantaneous rates of change by computing average rates of change over successively smaller intervals. As $\Delta t$ gets closer and closer to zero, we watch to see what value those average rates of change approach.

Yesterday in class Dr. MBK made you do some horrible algebra in order to alegbraically prove the sequence was going to the number we expected.

Today we are going to see how Python can help us avoid doing all this algebra. 

Remember that instead of choosing a particular value of $\Delta t$, we keep $\Delta t$ as a symbol and simplify the average rate of change algebraically. Then we plug in $\Delta t = 0$ and magically get a number. 

Let's see how Python can make all this happen using the sympy library.

```python
import sympy as sp

# Define delta_t as a mathematical symbol
delta_t = sp.symbols('delta_t')

# Define the function
P = lambda t: ???

# Point of interest
t0 = ???

# compute average rate of change over [t0, t0 + delta_t], keeping delta_t as symbol
average_rate = ???

# Simplify the average rate of change
simplified_rate = sp.simplify(average_rate)

print("The value we want to compute is P'(6.5) = (P(6.5 + delta_t) - P(6.5)) / delta_t")
print("The simplified expression is: ",simplified_rate)
```

```python
# What happens as delta_t approaches 0?
# name = simplified_rate.subs(delta_t, value of delta_t to plug in)
limit = simplified_rate.subs(delta_t, 0)

print("The limit as delta_t goes to 0 is:",limit)
```

```python
# Python can also evaluate the limit directly
# name = sp.limit(delta, function, value of delta_t to plug in)
python_limit = sp.limit(average_rate, delta_t, 0)

print("The limit as delta_t goes to 0 is:",python_limit)
```
